```python
# Financial Data Analysis of Microsoft, Apple, and Tesla (2023–2025)

## Objective
The objective of this analysis is to examine the financial performance of Microsoft, Apple, and Tesla using key financial metrics extracted from their 10-K reports. The analysis focuses on identifying trends in revenue, net income, assets, liabilities, and operating cash flow to support the development of an AI-powered financial chatbot.

## Methodology
- Extracted financial data from the SEC 10-K reports.
- Organized the data into an Excel spreadsheet.
- Imported the data into Python using pandas.
- Calculated year-over-year revenue growth and net income growth.
- Compared the financial performance of the three companies.
```

```python
# Findings and Insights

### Microsoft
- Revenue increased consistently from 2023 to 2025.
- Net income also increased steadily, indicating strong profitability.
- Operating cash flow showed continuous improvement, reflecting healthy business operations.

### Apple
- Apple generated the highest revenue among the three companies.
- Net income declined slightly in 2024 before recovering strongly in 2025.
- Operating cash flow remained consistently above $110 billion.

### Tesla
- Revenue remained relatively stable over the three years.
- Net income declined significantly in 2024 and again in 2025.
- Assets increased every year, showing continued investment and expansion despite lower profitability.

## Overall Conclusion

Microsoft demonstrated consistent financial growth across all years. Apple maintained the largest revenue and strong operating cash flow, while Tesla experienced declining profitability despite increasing assets. These trends provide valuable financial insights that can be incorporated into an AI-powered financial chatbot to answer user questions about company performance.
```
